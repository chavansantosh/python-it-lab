# Styled strip plot

```python
# style strip plot
plt.figure(figsize = (10,8))

# create the plot

sns.stripplot(data= tips,
              x = 'day',
              y = 'total_bill',
              hue = 'sex',
              palette =['green','blue'],
              size =5,
              alpha=0.7,
              jitter =True,
              dodge=True
             )

plt.legend(title="Gender")
plt.title("stripplot Example",fontsize =16, fontweight ='bold')
plt.xlabel("day")
plt.ylabel("total Bill")
plt.show()
```

# Box plot

```python
# what is boxplot ( box plot is five number of summary)

plt.figure(figsize = (10,8))

# create the advance box plot
sns.boxplot(data= tips,
            x = 'day',
            y = 'total_bill',
            hue = 'smoker',
            palette =['lightblue','lightgreen'],
            width= 0.7,
            linewidth = 1.5,
            flierprops ={
                        'marker': 'o',
                'markerfacecolor': 'red',
                'markeredgecolor': 'black',
                'markersize': 6
            },

            medianprops ={
                        'color': 'red',
                        'linewidth': 2.5,
    
            },
           )
```

```python
# create a anotations:
days =['Thur','Fri','Sat','Sun']

median = [tips[tips['day'] == d] ['total_bill'].median() for d in days]
```

# Relplot

# Relplot : is a figure level function for visualizing statistical relationshoips :

```python
# relplot with scatter:

sns.relplot(data=tips,
            x = 'total_bill',
            y = 'tip',
            hue = 'sex',
            size = 'size',
            style = 'smoker',
            palette =['blue','red'],
            sizes =(20,200),
            alpha =0.7,
            height =6,
            aspect =1.5
           )
```

## Line plot  with relplot

```python
sns.relplot(data= flights,
            x = 'year',
            y = 'passengers',
            hue = 'month',
            kind ='line',
            col ='month',  #one columns per month
            col_wrap=4 ,   # 4 columns per row
            palette ='tab20',
            height =3,
            aspect =1.3,
            linewidth=2
           )

plt.title("Line plot month wise")
plt.show()
```

# regplot

# regplot: show the scatter plot with a regression 

```python
sns.regplot(data=tips,
            x ='total_bill',
            y = 'tip',
            color = 'blue',
            scatter_kws={
                'alpha':0.9,
                'color':'blue'},
            line_kws={
                'color':'black',
                'linewidth':5,
                'alpha':0.9
            },
            
           )
```

# polynomial plot

```python
sns.regplot(data=tips,  # polynomial regression (order 2)
            x = 'size',
            y= 'total_bill',
            order = 2,
            color = 'red',
            scatter_kws={
                'alpha':0.5
            },
           )
```
