```python
import pandas as pd
df=pd.read_csv("C:\\Users\\MY PC\\OneDrive\\Desktop\\Data for pandas\\Task_dataset.csv")
df.head(5)
```

```python
""" 6  Data Type Audit
The company says
Age and AnnualIncome are stored incorrectly.
Tasks
Convert
Age → int
AnnualIncome → float
Balance → float
Target → string
Print information.
"""
import numpy as np
print(df["Age"].isna().sum())

df["Age"] = df["Age"].replace([np.inf, -np.inf], np.nan)

# Fill missing Age with median
df["Age"] = df["Age"].fillna(df["Age"].median())

# Convert to integer
df["Age"] = df["Age"].astype(int)

df["AnnualIncome"] = df["AnnualIncome"].astype(float)
df["Balance"] = df["Balance"].astype(float)
df["Target"] = df["Target"].astype(str)

print(df)
print(df.info())
```
