# Basic for loop

```python
#1.Print numbers from 1 to 10 using for loop.
for i in range(1,11):
    print(i,end=" ")
```

```python
#2.Take a number n from user and print sum of first n natural numbers using for loop.
sums = 0
num =int(input("please enter value:"))
for i in range(1,num+1):
    sums = sums + i
    
print("sums =",sums)
    
```

```python
#3.Print multiplication table of a given number using for loop.
num =int(input("please enter your number"))
for i in range(1,11):
    print(num,"x",i,"=",num * i)
```

```python
#4.Print all even numbers from 1 to 20 using for loop.
for i in range(1,21):
    if i % 2 ==0:
        print(i, end =" ")
```

```python
#5.Count how many numbers between 1 to 50 are divisible by 3 using for loop.
num = 0
for i in range(1,51):
    if i % 3 == 0:
        num+=1

print("divisible by 3:", num)
```

```python
#6.Print characters of a string one by one using for loop.
name = ["python"]
for i in name:   #iteration
    print(i)
    for j in i:
        print(j)
```

```python
#7.Print numbers from 50 to 30 in descending order using for loop.
for i in range(50, 29, -1):
    print(i,end = " ")
```

```python
#8.Find factorial of a number using for loop.
num =int(input("please enter your number:"))

fact = 1

for i in range(1, num + 1):
    fact = fact * i

print("Factorial =", fact)
```

# while loop

```python
#9.Print numbers from 1 to 15 using while loop.
i = 1
while i <= 15:
    print(i,end = " ")
    i+=1
```

```python
#10.Keep taking numbers from user until they enter 0, then print sum.
sum = 0

while True:
    num = int(input("Enter a number: "))

    if num == 0:
        break

    sum = sum + num

print("Sum =", sum)
```

```python
#11.Reverse a number using while loop.
num = 1234
rev = 0
while num >0:
    digit = num % 10
    rev = rev * 10 + digit
    num = num // 10
print(rev)   
```

```python
#12.Check if a number is prime using while loop.
num =int(input("enter number:"))
while num % 2 ==0:
    print("prime number")
    print(num)
```

```python
#13.Print Fibonacci series up to n terms using while loop.
n = int(input("Enter number: "))
a = 0
b = 1
i = 0

while i < n:
    print(a, end=" ")
    c = a + b
    a = b
    b = c
    i = i + 1
```

```python
#14.Keep asking password until correct one is entered.
password = input("Enter password: ")

while password != "1234":
    print("Wrong password")
    password = input("Enter password: ")

print("Correct password")
```
