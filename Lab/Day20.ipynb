```python
# 20. Create a custom exception called NegativeNumberError and raise it when user enters a negative number.
class NegativeNumberError(Exception):
    pass

num = int(input())

if num < 0:
    raise NegativeNumberError("Negative number is not allowed")
    
```

```python
#21. Create a custom exception InsufficientBalanceError for a banking program.
class InsufficientBalanceError(Exception):
    pass


balance = 1000
amount = int(input())

try:
    if amount > balance:
        raise InsufficientBalanceError("Insufficient balance")
    
    balance -= amount
    print("Withdrawal successful")
    print("Remaining balance:", balance)

except InsufficientBalanceError as e:
    print(e)
```
