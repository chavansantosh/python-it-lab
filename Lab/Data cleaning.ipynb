# 12.The Annual Income column contains values such as 650000 INR, ₹720000,
# 1.2L, and N/A. Clean the column and convert the valid values into numeric format.

```python
df['Annual_Income'].unique()
df['Annual_Income']=df['Annual_Income'].astype(str).str.strip()

df['Annual_Income'] = (
    df['Annual_Income']
    .str.replace('INR', '', regex=False)
    .str.replace('₹', '', regex=False)
    .str.replace(',', '', regex=False)
    .str.strip()
)
df['Annual_Income'] = pd.to_numeric(
    df['Annual_Income'], errors='coerce'
)
df['Annual_Income'].isna().sum()
```

# 27.Detect outliers in Annual Income using the IQR method.

```python
df['Annual_Income'] = pd.to_numeric(df['Annual_Income'], errors='coerce')


Q1 =df['Annual_Income'].quantile(0.25)
Q3 =df['Annual_Income'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[
    (df['Annual_Income'] < lower_limit) |
    (df['Annual_Income'] > upper_limit)
]
outliers
```
