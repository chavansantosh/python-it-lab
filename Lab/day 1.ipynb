# Here are 20 Practical Questions on Strings (Covering: Slicing, upper, lower, title, capitalize, isalpha, isdigit, isalnum, islower, find, index)

```python
# 1. Take a string from user and print the first 5 characters.
name=str(input("please enter your name:"))
print(name[:5])
```

```python
# 2. Take a string and print the last 4 characters using negative slicing.
name=str(input("enter a string"))
name[0:-4]
```

```python
# 3.Take a string and print characters from index 2 to 8 (excluding 8)
name = input("please enter a string")
print(name[2:8])
```

```python
# 4. Take a string and print every 2nd character of the string.
name=input("Enter a string")
name[::2]
```

```python
# 5. Take a full name (e.g. "Sumit Kumar Sharma") and print only the first name and last name separately using slicing.
name="sumit kumar sharma"

first_name = name[:5]

last_name = name[-6:]

print(first_name.title())
print(last_name.title())
```

```python
# 6. Take a string from user and print it in: • Uppercase • Lowercase • Title Case • Capitalized (only first letter capital)
name=str(input("enter a string:"))
print(name.upper())
print(name.lower())
print(name.title())
print(name.capitalize())
```

```python
#7. Take a sentence and convert it to title case. Then print the original and converted version.
name = "I Love Programming"

print(name)

print(name.title())
```

```python
# 8. Take a mixed case string and check if lower() and upper() methods change the string permanently or not.
name = "hI tHiS PAritOsH"
print(name)
print(name.lower())
print(name.upper())
print("string permanently :",name)
```

```python
# 13. Take a password as input and check: • It should be alphanumeric • It should be minimum 8 characters long Print "Valid Password" or
 #"Invalid Password".

name = input("password")
print("valid pasword" * (name.isalnum() and (len(name)>8)) or "invalid")
```

# find() and index() Questions

```python
# 14. Take a string and a character from user. Find the first occurrence of that character using find().
name =input("enter a string")
A =input("Enter a charecter") 
print(name.find(A))
```

```python
# 15. Take a string and a word. Use find() to check if the word exists. If not found, print -1.
name =input("Enter a string :")

A = input("Enter a word:")

print(name.find(A))
```

```python
# 16. Take a sentence and find the position of the word "Python" using index().
name = " I Love Python Language"

print(name.index("Python"))
```

```python
# 17. Take a string. Use both find() and index() to search for a character that does not exist. Observe the difference.
name = "Hi I am Santosh and I Love programming"

print(name.find("papa"))
print(name.index("papa"))
```

```python
#18. Take your full name. • Convert it to uppercase • Find the position of first space • Print the first name and last name separately.
name = "Santosh Narayan Chavan"

print(name.upper())
print(name.find(" "))

first_name = name[:name.find(" ")]
last_name = name[name.rfind(" ") +1:]
print(first_name)
print(last_name)
```
