# 2. Plot two lines representing Product A and Product B sales for 6 months. Add
# a legend and different line styles.

```python

X = ['January','February','March','April','May','June']

Product_A = [21000,23540,32100,40040,42651,45000]
Product_B = [18000,25000,28000,35000,38000,42000]

plt.figure(figsize=(10, 6))

plt.plot(X, Product_A, marker='o', linestyle='-', label='Product A')
plt.plot(X, Product_B, marker='*', linestyle='--', label='Product B')

plt.xlabel('Month')
plt.ylabel('Sales')
plt.title('Product A $ Product B Sales')

plt.grid()
plt.legend()

plt.show()
```

# 8. Create a plot for:
# where x ranges from -10 to 10. Add a horizontal line at y=0.

```python
x = np.linspace(-10, 10, 100)
y = x ** 2

plt.figure(figsize=(10, 6))

plt.plot(x, y, label='y = x²')

plt.axhline(y=0, label='y = 0')

plt.xlabel('X')
plt.ylabel('Y')
plt.title('Plot of y = x²')

plt.grid()
plt.legend()

plt.show()
```

# 9. Plot three mathematical functions on the same graph:
# Add different colors, markers, line styles, and a legend.

```python
x = np.linspace(-10, 10, 100)

y1 = x ** 2
y2 = x
y3 = x ** 3

plt.figure(figsize=(10, 6))

plt.plot(x, y1, color='red',marker='o',linestyle='-',label='y = x²')
plt.plot(x, y2, color='blue',marker='s',linestyle='--',label='y = x')
plt.plot(x, y3, color='green',marker='^',linestyle=':',label='y = x³')

plt.xlabel('X')
plt.ylabel('Y')
plt.title('Mathematical Functions')

plt.grid()
plt.legend()

plt.show()
```

# 10.Create a 2×2 subplot containing:
# o Line chart
# o Bar chart
# o Scatter plot
# o Histogram

```python
x = np.arange(1, 7)
y = [20,35,25,45,40,55]

categories = ['A','B','C','D','E']
values = [25, 40, 30, 50, 35]

scatter_x = [1,2,3,4,5,6,7,8]
scatter_y = [10,25,20,40,35,50,45,60]

random_values = np.random.randn(100)

plt.figure(figsize=(12, 8))

# Line chart
plt.subplot(2,2,1)
plt.plot(x, y, marker='o')
plt.title('Line Chart')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.grid()

# Bar chart
plt.subplot(2,2,2)
plt.bar(categories, values)
plt.title('Bar Chart')
plt.xlabel('Category')
plt.ylabel('Value')

# Scatter plot
plt.subplot(2,2,3)
plt.scatter(scatter_x, scatter_y)
plt.title('Scatter Plot')
plt.xlabel('X')
plt.ylabel('Y')
plt.grid()

# Histogram
plt.subplot(2,2,4)
plt.hist(random_values, bins=10)
plt.title('Histogram')
plt.xlabel('Values')
plt.ylabel('Frequency')
plt.grid()

plt.tight_layout()
plt.show()
```
