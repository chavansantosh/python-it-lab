```python
# 1. Create a list original = [1, 2, [3, 4], 5]. Make a shallow copy named shallow and change
# shallow[2][0] = 99. Print both lists and explain the result.
import copy
list1 = [1, 2, [3, 4], 5]
list2 =copy.copy(list1)
list2[2][0]=99
print("list1 :",list1)
print("list2 :",list2)
```

```python
# 2. Do the same as above but use deepcopy instead of shallow copy. Show the difference.
import copy
list1 = [1, 2, [3, 4], 5]
list2 =copy.deepcopy(list1)
list2[2][0]=99

print("list1 :",list1)
print("list2 :",list2)
```

```python
#9. (Hard) Create a function that detects whether a copy is shallow or deep by modifying a
# nested element and checking the original.
import copy

def check_copy(copied):
    copied[1][0] = 10

    if copied[1][0] == 10:
        print("Shallow copy")
    else:
        print("Deep copy")

original = [1, [2, 3, 4], 5]

# Shallow copy
shallow = copy.copy(original)
check_copy(shallow)

# Reset original
original = [1, [2, 3, 4], 5]

# Deep copy
deep = copy.deepcopy(original)

if deep[1][0] == original[1][0]:
    deep[1][0] = 10
```

```python
#10. (Hard) Create a class Student with attributes name and marks (list). Demonstrate what
# happens when you do shallow copy vs deep copy of a Student object.
import copy


class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks


# Original Student object
student1 = Student("Santosh", [85, 54, 89])

# Shallow copy
student2 = copy.copy(student1)

student2.marks[0] = 100

print("After Shallow Copy")
print("Student 1 marks:", student1.marks)
print("Student 2 marks:", student2.marks)


# Create original object again
student1 = Student("Santosh", [85, 54, 89])

# Deep copy
student3 = copy.deepcopy(student1)

student3.marks[0] = 100

print("\nAfter Deep Copy")
print("Student 1 marks:", student1.marks)
print("Student 3 marks:", student3.marks)
```
