# 15.Convert Dep_Time into datetime format and extract the departure hour into Dep_Hour.

```python
df['Dep_Time'] = pd.to_datetime(df['Dep_Time'],format='%H:%M',errors = "coerce")

df['Dep_Hour'] = df['Dep_Time'].dt.hour

df[['Dep_Time', 'Dep_Hour']].head()
```
