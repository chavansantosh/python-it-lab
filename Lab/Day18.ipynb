# Method Overloading Questions (1-7)

```python
"""
 1. Create a Calculator class that can add 2 numbers or 3 numbers using the same method name.
        # Concept
 Method Overloading (Python uses default arguments or *args to achieve overloading.)
 What to do
 • Create a class Calculator.
 • Create one method add().
 • The method should work for either 2 numbers or 3 numbers.
"""
class Calculator:
    def add(self,a,b,c = 0):
        return a + b + c

cal = Calculator()

print(cal.add(20,85))
print(cal.add(5,3,5))
```

```python
""""
2. Create an Area class that calculates the area of a square and rectangle using the same
     # method.
 Concept
 Method Overloading using default arguments.
 What to do
 • One parameter → Square.
 • Two parameters → Rectangle.
 """
class Area:
    def calculate(self,length,breadth=None):
        if breadth is None:
            return length * length
        else:
            return length * breadth
A1 = Area()
print("Area of square ",A1.calculate(5))

print("Area of Rectangle ",A1.calculate(5,20))       
        
```

```python
""""
3. Create a Student class with a details() method that accepts different numbers of
    # arguments.
 Concept
 Method Overloading
 What to do
 The method should work with:
 • Name
 • Name + Age
• Name + Age + City
""""
class Student:
    def details(self,name,age = None,city = None):
        print("Name :" ,name)

        if age is not None:
            print("Age :",age)

        if city is not None:
            print("City :",city)

# creating object:
s = Student()
print(s.details("santosh",24))
print(s.details("paritosh",25,"hyd"))
```

```python
""""
4. Create a MathOperations class that multiplies any number of values.
        # Concept
 Method Overloading using *args
 What to do
 Multiply:
 • 2 numbers
 • 3 numbers
 • 4 numbers
 • More numbers
 """
class MathOperation:
    
    def multiply(self,*number):
        result = 1
        
        for i in number:
            result = result * i
            
        return result

A = MathOperation()

print(A.multiply(2,3))
print(A.multiply(2,3,6))
print(A.multiply(2,7,1,5))
                      
```

```python
"""
5. Create a Display class that prints different messages based on the argument type.
    # Concept
 Method Overloading + Type Checking
 What to do
Accept:
 • Integer
 • String
 • List
 """
class Display:
    def show(self,value):

        if isinstance(value,int):
            print("Integer value:",value)

        elif isinstance(value,str):
            print("String value :",value)

        elif isinstance(value,list):
            print("List value :",value)
        else:
            print("No value is found")

obj = Display()

obj.show(25)
obj.show("santosh")
obj.show([1,2,3,4])
            
    
```

```python
"""
6. Create a Volume class that calculates the volume of a cube and cuboid.
Concept
Method Overloading
What to do
• One parameter → Cube
• Three parameters → Cuboid
"""
class Volume:
    def calculate(self,a, b = None,c = None):
        if b and c is None:
            print("Cube is :",a**3)
        else:
            print("Cuboid is :",a**3)

obj = Volume()
obj.calculate(2)
obj.calculate(5)
```

```python
"""
7. Create a PrintData class that prints one or many values.
Concept
Method Overloading using *args
What to do
Print:
• One value
• Multiple values
"""
class printData:
    def print_data(self,*args):

        for value in args:
            print(value)

obj = printData()

obj.print_data(20)
obj.print_data("paritosh",51,"hyd",123456)
```

# Method Overriding Questions (8-14)

```python
"""
8. Create an Animal class with sound(). Override it in Dog and Cat.
Concept
Method Overriding
What to do
• Animal → Generic sound
• Dog → Bark
• Cat → Meow
"""
class Animal:
    def sound(self):
        return "sound"

class Dog (Animal):
    def sound(self):
        print("Dog is barking")

class Cat (Animal):
    def sound(self):
        print("Cat is meoww")
animal = Animal()
dog = Dog()
cat = Cat()

animal.sound()
dog.sound()
cat.sound()
```

```python
"""
9. Create Vehicle, Car, and Bike classes.
Concept
Method Overriding
What to do
Override the start() method.
How to solve
1. Create Vehicle.
2. Add method start().
3. Create Car.
4. Create Bike.
5. Override start() in both child classes.
"""
class Vehicle:
    def start(self):
        print("Vehicle is starting")

class Car(Vehicle):
    def start(self):
        print("Car is starting")

class Bike(Vehicle):
    def start(self):
        print("Bike is starting")

vehicle = Vehicle()
car = Car()
bike = Bike()

vehicle.start()
car.start()
bike.start()
```

```python
"""
10. Create an Employee class and override calculate_salary() in Manager.
Concept
Method Overriding
What to do
• Employee → Normal salary
• Manager → Salary + Bonus
"""
class Employee:
    def __init__(self,salary):
        self.salary = salary

    def calculate_salary(self):
        return  self.salary

class Manager(Employee):
    def __init__(self,salary,bonus):
        super().__init__(salary)
        self.bonus = bonus

    def calculate_salary(self):
        return self.salary + self.bonus

emp = Employee(5000)
manager = Manager(70000,5000)

print(emp.calculate_salary())
print(manager.calculate_salary())
```

```python
"""
11. Create Shape, Circle, and Rectangle classes.
Concept
Method Overriding
What to do
Override the area() method.
"""
class shape:
    def area(self):
        return "Area"

class Circle(shape):
    def __init__(self,radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius 

class Rectangle(shape):
    def __init__(self,length,width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

shape = shape()
circle = Circle(5)
rectangle = Rectangle(8, 4)

print(shape.area())
print(circle.area())
print(rectangle.area())
```

```python
"""
12. Create BankAccount, SavingsAccount, and CurrentAccount.
Concept
Method Overriding
What to do
Override the interest() method.
"""
class BankAccount:
    def interest(self):
        return "Interest"

class SavingAccount(BankAccount):
    def __init__(self,balance):
        self.balance = balance

    def interest(self):
        return self.balance * 4

class CurrentAccount(BankAccount):
    def __init__(self,balance):
        self.balance = balance

    def interest(self):
        return 0

bankaccount = BankAccount()
savingaccount = SavingAccount(5000)
currentaccount = CurrentAccount(70000)

print(savingaccount.interest())
print(currentaccount.interest())
```
