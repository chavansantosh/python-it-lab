```python
# 16.Generator with Lists
# Given the following list: numbers = [10, 15, 20, 25, 30, 35]
# Create a generator that yields only the even numbers.
numbers = [10, 15, 20, 25, 30, 35]
def even_numbers():
    for i in numbers:
        if i % 2 ==0:
            yield i
    
A = even_numbers()
for i in A:
    print(i)
```

```python
# Problem 21
# Create a custom iterator that prints the first 15 natural numbers.
class NaturalNumber:
    def __init__(self):
        self.num = 1
        
    def __iter__(self):
        return self
    def __next__(self):
        if self.num <= 15:
             value = self.num
             self.num += 1
             return value
            
        else:
            raise StopIteration

A = NaturalNumber()

for i in A:
    print(i)
    
```

```python
# Problem 22
# Create a custom iterator that prints all multiples of 3 up to 30.
class numbers:
    def __init__(self):
        self.num = 3

    def __iter__(self):
        return self

    def __next__(self):
        if self.num <= 30:
            value = self.num
            self.num += 3
            return value
        else:
            raise StopIteration

A = numbers()

for i in A:
    print(i)
        
```

```python
# Problem 24
# Create a custom iterator that prints all vowels from the string below.
# text = "Artificial Intelligence"
class Vowels:
    def __init__(self):
        self.name = "Artificial Intelligence"
        self.index = 0

    def __iter__(self):
        return self

    def __next__(self):
        while self.index < len(self.name):
            ch = self.name[self.index]
            self.index += 1

            if ch.lower() in "aeiou":
                return ch

        raise StopIteration


A = Vowels()

for i in A:
    print(i)
```

```python
# 25. Create a custom iterator that prints the Fibonacci sequence up to 10 terms.
class Fibonacci:
    def __init__(self):
        self.a = 0
        self.b = 1
        self.count = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.count < 10:
            value = self.a
            self.a, self.b = self.b, self.a + self.b
            self.count += 1
            return value
        else:
            raise StopIteration


A = Fibonacci()

for i in A:
    print(i)
```

```python
# Problem 26
# Create a generator that yields the Fibonacci sequence up to 15 terms.
class Fibonacci:
    def sequence(self):
        a = 0 
        b = 1

        for i in range(16):
            yield a

            a, b = b, a + b

A = Fibonacci()
for i in A.sequence():
    print(i)
```

```python
# Problem 27
# Create a generator that yields all prime numbers between 1 and 100.
def prime_no():
    for num in range(2,101):
        prime = True

        for i in range(2,num):
            if num % i == 0:
                prime = False
                break

        if prime:
            yield num

for i in prime_no():
    print(i)
    
```
