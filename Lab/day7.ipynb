```python
#8.take three side of triangle.first check if they form of valid triangle,then check if its is Equilateral,isosceles,or scalene
A = int(input("enter first side value"))
B = int(input("enter second side value"))
C = int(input("enter third side value"))
if (A + B >C) and (A + C >B) and (B + C >A):
        print("Valid Triangle")
        if A == B == C:
            print("Equilateral Triangle")
        elif A == B or B == C or A == C:
            print("Isosceles Triangle")
        else:
            print("Scalene Triangle")
else: 
    print("Invalid Triangle")
```

```python
# 10.Take a character. If it is a letter, check if it is vowel or consonant. If it is a digit, print "Digit". Else print "Special Character".
A = input("Enter a character: ")
if A.isalpha():
    if A.lower() in "aeiou":
        print("Vowel")
    else:
        print("Consonant")
elif A.isdigit():
    print("Digit")
else:
    print("Special Character")
```

```python
# 11.Take a year as input. Check if it is a leap year using nested conditions.
year = int(input("Enter a year: "))

if year % 4 == 0:
    if year % 100 == 0:
        if year % 400 == 0:
            print("Leap Year")
        else:
            print("Not a Leap Year")
    else:
        print("Leap Year")
else:
    print("Not a Leap Year")
```

```python
#13.Take three numbers. Check if all three are equal, or any two are equal using nested if.
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
c = int(input("Enter third number: "))

if a == b == c:
    print("All three numbers are equal")
else:
    if a == b or a == c or b == c:
        print("Any two numbers are equal")
    else:
        print("All numbers are different")
```
