# 12.Find all unique values in the gender column and convert them into consistent
# categories such as Male and Female.

```python
import pandas as pd 

df = pd.read_csv("C:\\Users\\MY PC\\Downloads\\customer_data - customer_data.csv")
df.head(2)
```

```python
df['gender'].unique()
df['gender'] = df['gender'].str.strip().str.title()
df['gender'].unique()
df['gender'] = df['gender'].str.strip().str.upper().replace({
    "M":"Male",
    "F":"Female"
})
df['gender']
df['gender'].unique()
```
