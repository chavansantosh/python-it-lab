```python
# multi-figure methods
X = np.linspace(0,10,100)

# create the multiplot figure
fig,axes = plt.subplots( nrows = 2,   
                         ncols = 2,
                         figsize=(10,6)
)



# fig
fig.set_size_inches(7,6) # resize in inchess
fig.set_facecolor("yellow") # change the background color
fig.suptitle("Figure title",fontsize=17,
              fontweight='bold')

# plot in subplot:
axes[0,0].plot(X,np.sin(X),'blue')
axes[0,0].set_title("Sine plot")

axes[0,1].plot(X,np.cos(X),'black')
axes[0,1].set_title("Cos plot")


axes[1,0].plot(X,np.tanh(X),'green')
axes[1,0].set_title("Tanh plot")

axes[1,1].plot(X,X**2,'Red')
axes[1,1].set_title("Square plot")

plt.show
```

```python
# multi-figure eith axes method
X = np.linspace(0,10,100)

# create the multiplot figure
fig,axes = plt.subplots( nrows = 2,   
                         ncols = 2,
                         figsize=(36,24)
)



# fig
fig.set_size_inches(7,6) # resize in inchess
fig.set_facecolor("yellow") # change the background color
fig.suptitle("Figure title",fontsize=17,
              fontweight='bold')

# plot in sub
axes[0,0].plot(X,np.sin(X),'blue',linewidth = 4)
axes[0,0].set_title("Sine plot")
axes[0,0].set_xlabel("X axis label",fontsize=12)
axes[0,0].set_ylabel("y axis label",fontsize=12)
axes[0,0].set_xlim(0,10)
axes[0,0].set_ylim(0,2)   
axes[0,0].legend(['Sine plot'])
axes[0,0].grid(True,linestyle='--',alpha=0.9)

axes[0,1].plot(X,np.cos(X),'black',linewidth = 4)
axes[0,1].set_title("Cos plot")
axes[0,1].set_xlabel("X axis label",fontsize=12)
axes[0,1].set_ylabel("y axis label",fontsize=12)
axes[0,1].set_xlim(0,10)
axes[0,1].set_ylim(0,2)
axes[0,1].legend(['Cos plot'])
axes[0,1].grid(True,linestyle='--',alpha=0.9)

axes[1,0].plot(X,np.tanh(X),'green',linewidth = 4)
axes[1,0].set_title("Tanh plot")
axes[1,0].set_xlabel("X axis label",fontsize=12)
axes[1,0].set_ylabel("y axis label",fontsize=12)
axes[1,0].set_xlim(0,10)
axes[1,0].set_ylim(0,2)
axes[1,0].legend(['Tanh plot'])
axes[1,0].grid(True,linestyle='--',alpha=0.9)

axes[1,1].plot(X,X**2,'Red',linewidth = 4)
axes[1,1].set_title("Square plot")
axes[1,1].set_xlabel("X axis label",fontsize=12)
axes[1,1].set_ylabel("y axis label",fontsize=12)
axes[1,1].set_xlim(0,10)
axes[1,1].set_ylim(0,2)
axes[1,1].legend(['Square plot'])
axes[1,1].grid(True,linestyle='--',alpha=0.9)

plt.show
```

```python
# Multiplot  figure with Axes methods 

X =  np.linspace(0,10,100)

#create the multiplot figure

fig,axes = plt.subplots(2,3,
                        dpi = 60,
                       figsize = (15,12))


# figure method

fig.set_facecolor("lightgreen")  # change background color
fig.suptitle("Figure Title",
              fontsize=26,  # font size figure title
             fontweight='bold')
 


# Plot in each subplot 
axes[0,0].plot(X, np.sin(X), 'olive', linewidth=4) 
axes[0,0].set_title("Sine plot",fontsize=16)
axes[0,0].set_xlabel("X Axis Label", fontsize=12) 
axes[0,0].set_ylabel("Y Axis Label", fontsize=12) 
axes[0,0].set_xlim(0,12)
axes[0,0].set_ylim(-1.5,1.5)
axes[0,0].legend(["Sine Waves"])
axes[0,0].grid(True, linestyle='--', alpha=0.9) 
axes[0,0].axhline(y=0, color='black', 
           linewidth=1, linestyle='-')


# Plot in each subplot 
axes[0,1].plot(X, np.cos(X), 'slategray',linewidth=4) 
axes[0,1].set_title("Cos plot",fontsize=16) 
axes[0,1].set_xlabel("X Axis Label", fontsize=12) 
axes[0,1].set_ylabel("Y Axis Label", fontsize=12)
axes[0,1].set_xlim(0,12)
axes[0,1].set_ylim(-1.5,1.5)
axes[0,1].legend(["Cose Waves"])
axes[0,1].grid(True, linestyle='--', alpha=0.9) 
axes[0,1].axhline(y=0, color='black', 
           linewidth=1, linestyle='-')

# Plot in each subplot 
axes[0,2].bar([1,2,3,4,6], [4,5,6,2,7], color='green')  
axes[0,2].set_title("Bar plot",fontsize=16) 
axes[0,2].set_xlabel("X Axis Label", fontsize=12) 
axes[0,2].set_ylabel("Y Axis Label", fontsize=12)
axes[0,2].set_xlim(0,8)
axes[0,2].legend(["Cose Waves"])
axes[0,2].grid(True, linestyle='--', alpha=0.9) 



# Plot in each subplot 
axes[1,0].plot(X, np.tanh(X), 'orange',linewidth=4) 
axes[1,0].set_title("Tanh plot",fontsize=16) 
axes[1,0].set_xlabel("X Axis Label", fontsize=12)
axes[1,0].set_ylabel("Y Axis Label", fontsize=12)
axes[1,0].set_xlim(0,10)
axes[1,0].set_ylim(0,1.25)
axes[1,0].legend(["Tanh Waves"])
axes[1,0].grid(True, linestyle='--', alpha=0.9) 
axes[1,0].axhline(y=0, color='black', 
           linewidth=1, linestyle='-')

# Plot in each subplot 
axes[1,1].plot(X, X**2, 'tan',linewidth=4) 
axes[1,1].set_title("Squred plot",fontsize=16)
axes[1,1].set_xlabel("X Axis Label", fontsize=12) 
axes[1,1].set_ylabel("Y Axis Label", fontsize=12)
axes[1,1].set_xlim(0,10)
axes[1,1].set_ylim(0,100)
axes[1,1].legend(["Squared Waves"])
axes[1,1].grid(True, linestyle='--', alpha=0.9) 


# Plot in each subplot 
axes[1,2].scatter(X[:50], np.sin(X[:50]),color='purple', s=20) 
axes[1,2].set_title("Squred plot",fontsize=16)
axes[1,2].set_xlabel("X Axis Label", fontsize=12) 
axes[1,2].set_ylabel("Y Axis Label", fontsize=12)
axes[1,2].set_xlim(0,5)
axes[1,2].set_ylim(0,1)
axes[1,2].legend(["Squared Waves"])
axes[1,2].grid(True, linestyle='--', alpha=0.9) 

plt.show()
```

# how to add legends

```python
# create the data :
X = np.linspace(0,15,200)
y = np.sin(X)

#create the figure
plt.figure(figsize = (10,8))

# plot the data
plt.plot(X ,y,color = "blue",
         marker = "o",
         markerfacecolor = "orange",
         markeredgecolor = "black",
         markersize = 3
        )
# add legend:
plt.legend(["sine"],
           loc ='lower right',
           fontsize = 15,
           title = "info",
           title_fontsize = 14,
           edgecolor = "black",
           facecolor = "lightblue",
           shadow = True
          )

plt.title("Legend Graph")

plt.show()
```

```python
# plot1 : line plot:
months = ["jan","feb","mar","apr","may","jun","jul","aug","sep","oct","nov","dec"]
product_a =[120,135,148,162,155,178,195,210,198,220,235,250]
product_b =[80,95,88,105,112,120,135,128,142,155,160,175]
product_c =[60,65,70,75,85,90,95,100,108,115,120,130]

# make the plot
X = range(len(months))

# make the sub plots
fig, ax=plt.subplots(2,2,
                     figsize=(12,6))

# make the plot
ax[0,0].plot(X,product_a,
       color = 'purple',
       linewidth = 2,
       marker = "o",
       markersize = 6,
       label = "Product A")
ax[0,0].set_title("Product A")
ax[0,0].legend(["Product A"])

ax[0,1].plot(X,product_b,
       color = 'lightgreen',
       linewidth = 2,
       marker = "o",
       markersize = 6,
       label = "Product B")
ax[0,1].set_title("Product B")
ax[0,1].legend(["Product B"])

ax[1,0].plot(X,product_c,
       color = 'lightblue',
       linewidth = 2,
       marker = "o",
       markersize = 6,
       label = "Product C")
ax[1,0].set_title("Product C")
ax[1,0].legend(["Product C"])

ax[1,1].plot(product_a,product_b,
       color = 'brown',
       linewidth = 2,
       marker = "o",
       markersize = 6,
       label = "Product A"
       )
ax[1,1].set_title("Product B vs A")
ax[1,1].legend(["Product B vs A"])

plt.show()
```

```python
# Department salary data 
departments = ['Engineering', 'Marketing', 
               'Sales', 'HR', 'Finance'] 
avg_salary  = [95000, 75000, 70000, 
               65000, 85000] 
max_salary  = [150000, 110000, 120000, 
               90000, 130000] 

# make the subplots
fig,ax = plt.subplots(1,2,figsize = (12,8))

x = np.arange(len(departments)) 
width = 0.35     # bar width 
# make the plot 

# Two sets of bars 
ax[0].bar(x - width/2, 
               avg_salary, 
               width, 
               label='Average Salary', 
               color='purple', 
               edgecolor='black', 
               linewidth=0.5)
ax[0].set_ylim(10000,90000)

ax[1].bar(x + width/2, 
               avg_salary, 
               width, 
               label='Average Salary', 
               color='lightgreen', 
               edgecolor='black', 
               linewidth=0.5)
```
