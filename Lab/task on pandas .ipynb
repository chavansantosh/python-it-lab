```python
# 3. Take numbers from user until they enter negative number. Calculate sum and average using while loop.
total = 0
count = 0

while True:
    num =int(input("please enter a number"))
    if num < 0:
        break

    total += num
    count += 1

if count > 0:
    average = total / count
    print("sum:",total)
    print("average",average)
else:
    print("It a Negative number")
```

# OOP + Encapsulation (10-14)

```python
# 10. Create a class Student with private attribute __marks. Add getter and setter with validation (0–100).
class Student:
    def __init__(self, marks):
        self.__marks = marks

    # Getter
    def get_marks(self):
        return self.__marks

    # Setter
    def set_marks(self, marks):
        if 0 <= marks <= 100:
            self.__marks = marks
            print("Marks updated")
        else:
            print("Invalid marks")

student = Student(85)

print("Current Marks:",student.get_marks())

student.set_marks(95)
print("Updated Marks:",student.get_marks())
```

```python
# 11. Create class BankAccount with private __balance. Add deposit and withdraw
# methods. Raise custom exception if balance becomes negative.    
class BankAccount:
    def __init__(self,balance):
        self.__balance = balance

    def deposit(self,amount):
        self.__balance += amount
        print("amount deposited :",amount)

    def withdraw(self,amount):
        if  self.__balance - amount < 0:
            raise InsufficientBalanceError("insufficient balance")
        else:
            self.__balance -= amount
            print("amount withdrawn",amount)

    def get_balance(self):
        return self.__balance

account = BankAccount(5000)

print("Balance :",account.get_balance())

account.deposit(7000)
print("Deposit :",account.get_balance())

try:
    account.withdraw(7000)
except InsufficientBalanceError as e:
    print(e)

print("Balance:", account.get_balance())
```

```python
# 13. Create class Product with private __price. Add getter and setter. Setter should
 # not allow price less than 100.
class Product:
    def __init__(self,price):
        self.__price = price

    def get_Product(self):
        return self.__price

    def set_Product(self,price):
        if price < 100:
            print("amount should be more than 100")
        else:
            self.__price = price
            print("amount updated")

items = Product(101)

print("current price:",items.get_Product())

items.set_Product(124)
print("updated price:",items.get_Product())

items.set_Product(99)
print(items.get_Product())
```

```python
# 14. Create class Rectangle with private length and width. Add method to calculate
# area and perimeter.
class Rectangle:
    def __init__(self,length,width):
        self.__length = length
        self.__width = width

    def area(self):
        return  self.__length * self.__width

    def perimeter(self):
        return 2 *( self.__length + self.__width )

        print("area:",area)
        print("perimeter:",perimeter)

ret = Rectangle(5,2)
print(ret.area())
print(ret.perimeter())

        
        
```
