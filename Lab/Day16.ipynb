# Basic Hierarchical Inheritance (1–12)

```python
# 1. Person → Student, Teacher
# Create a parent class Person with attributes name and age. Create two child classes Student and Teacher that inherit from Person.
      # Concept
#• Hierarchical Inheritance
#• Parent attributes inherited by multiple child classes.
# 1. Create a parent class Person.
# 2. Add name and age using a constructor.
# 3. Create Student(Person).
# 4. Create Teacher(Person).
# 5. Create objects of both classes.
# 6. Show that both children can access name and age.
class person:
    def __init__(self,name,age):
        self.name = name
        self.age = age

#class child 1:
class Student(person):
    pass

#child class 2:
class Teacher(person):
    pass

# creating a object:
student = Student("santosh",24)
teacher = Teacher("paritosh",25)

#display info
print("Student details")
print("Name:",student.name)
print("Age :",student.age)

print("Teacher details")
print("Name :",teacher.name)
print("Age :",teacher.age)
```

```python
#2. Animal → Dog, Cat
#Define class Animal with method eat(). Create subclasses Dog and Cat.
     #Concept
#• Method inheritance.
#How to Solve
# 1. Create parent class Animal.
# 2. Write method eat().
# 3. Create child classes Dog and Cat.
# 4. Create objects.
# 5. Call eat() using both objects.
class Animal:
    
    def eat(self):
        return "Animal is eating"

class Dog(Animal):
    pass

class Cat(Animal):
    pass

#creating object:
dog = Dog()
cat = Cat()

#calling:
print("Dog :",dog.eat())
print("Cat :",cat.eat())
```

```python
# 3. Vehicle → Car, Bike
# Create class Vehicle with method start(). Create subclasses Car and Bike.
# Concept
# • One parent method shared by multiple children.
        # How to Solve
# 1. Create Vehicle.
# 2. Add start().
# 3. Inherit it in Car.
# 4. Inherit it in Bike.
# 5. Call start() from both child objects.
class Vehicle:

    def start(self):
        return f"Vehicle is running"

#child class 1:
class Car(Vehicle):
    pass

#child class 2:
class Bike(Vehicle):
    pass

#creating object:
car = Car()
bike = Bike()

#calling
print("Car :",car.start())
print("Bike :",bike.start())
```

```python
#4. Shape → Circle, Rectangle
     # Question
# Define class Shape with method area(). Create subclasses Circle and Rectangle.
        #Concept
# • Parent provides common method.
# • Children may later override it.
# How to Solve
# 1. Create Shape.
# 2. Add area().
# 3. Create Circle.
# 4. Create Rectangle.
# 5. Call area().
class Shape:
    def area(self):
        return f"Area of Shape"
        
# child class 1:
class Circle(Shape):
    def area(self):
        return "Area = 3.14 * r * r"

# child class 2:
class Rectangle(Shape):
    def area(self):
        return "Area = length * width"

# creating object:
circle = Circle()
rectangle = Rectangle()

print(circle.area())
print(rectangle.area())
```

```python
# 5. Employee → Manager, Developer
         #Question
# Create class Employee with attributes emp_id and salary. Create subclasses Manager and
# Developer.
# Concept
#• Attribute inheritance.
        #How to Solve
# 1. Create parent constructor.
# 2. Store emp_id and salary.
# 3. Create two child classes.
# 4. Show inherited attributes.

class Employee:
    def __init__(self,emp_id,salary):
        self.emp_id = emp_id
        self.salary = salary


# child class 1:
class Manager(Employee):
    pass

#child class 2:
class Devloper(Employee):
    pass

# creating object:
manager = Manager(101,59786)
devloper = Devloper(102,856868)

print("Manager ID",manager.emp_id)
print("Manager salary",manager.salary)

print("Devloper ID",devloper.emp_id)
print("Devloper salary",devloper.salary)
```
