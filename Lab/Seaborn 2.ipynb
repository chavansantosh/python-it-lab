# B. Tips Dataset — 15 Questions

```python
import seaborn as sns
import numpy as np
import pandas as pd
df = pd
import matplotlib.pyplot as plt
```

# Load the dataset:

```python
tips = sns.load_dataset("tips")
tips
```

## 2. Total Bill by Day
## Use catplot() to compare total_bill across:
## • Thur
## • Fri
## • Sat
## • Sun
## Choose an appropriate kind.

```python
sns.catplot(data = tips,
            x = "day",
            y = "total_bill",
            kind = "point"
           )
plt.title("Total Bill by Day")
plt.xlabel("day")
plt.ylabel("total_bill")
plt.show()
```

# 3. Tip by Day
# Create a categorical visualization showing tip distribution across different days.

```python
sns.catplot(data = tips,
            x = "day",
            y = "tip",
            kind = "swarm"
           )
plt.title("Tip by Day")
plt.xlabel("day")
plt.ylabel("tip")
plt.show()
```

## 4. Total Bill by Gender
## Compare total_bill between:
## • Male
## • Female
## Use catplot().

```python
sns.catplot(data = tips,
            x = "sex",
            y = "total_bill",
            kind = "swarm"
           )
plt.title("Total Bill by Gender")
plt.xlabel("sex")
plt.ylabel("total_bill")
plt.show()
```

# 5. Tip by Gender
# Create a categorical plot comparing tips given by male and female customers.

```python
sns.catplot(data = tips,
            x = "sex",
            y = "tip",
            kind = "violin",
            height=5,
            aspect=1.2,
            palette='Set2'
           )
plt.title("Tip by Gender")
plt.xlabel("sex")
plt.ylabel("tip")
plt.show()
```

## 6. Total Bill by Smoker Status
## Compare the distribution of total_bill between:

## • Non-smokers
## Use hue="smoker" where appropriate.

```python
sns.catplot(data=tips,
            x = 'smoker',
            y = 'total_bill',
            hue = 'smoker',
            kind = 'strip'
        
           )
plt.title("Total Bill by Smoker Status")
plt.xlabel("total_bill")
plt.ylabel("smoker")
plt.show()
```

## 14. Multi-Panel Tips Analysis
## Create a catplot() with:
## • X → day
## • Y → total_bill
## • Hue → sex
## • Columns → time
## Choose an appropriate kind.
## Then compare lunch vs dinner.

```python
sns.catplot(data=tips,
            x ='day',
            y ='total_bill',
            hue ='sex',
            col='time',
            kind ='bar',
            
           )
plt.title("   Multi-Panel Tips Analysis",y=1.05)
plt.xlabel("day")
plt.ylabel("total_bill")

plt.show()
```

## 15. Final Seaborn catplot() Challenge
## Using the Tips dataset, create a professional categorical visualization with:
## x="day"
## y="total_bill"
## hue="sex"
## col="time"
## Choose the most appropriate kind.
## Customize:
## • palette
## • height
## • aspect
## • col_wrap if appropriate
## • Title
## • Axis labels
## • Legend

```python
sns.catplot(
    data=tips,
    x='day',
    y='total_bill',
    hue='sex',
    col='time',
    kind='bar',
    palette=['blue','red'],
    height=5,
    aspect=1.2
)

plt.subplots_adjust(top=0.85)
plt.suptitle("Multi-Panel Tips Analysis")

plt.show()
```
