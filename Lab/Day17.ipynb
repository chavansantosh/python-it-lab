```python
# 1. Create a hybrid inheritance structure with classes A → B and C → D (where D inherits from both B and C).
# Concept
# Hybrid Inheritance (Combination of Single + Multiple Inheritance)
    # What to do
# • Create parent class A.
# • Create class B inheriting from A.
# • Create another class C.
# • Create class D inheriting from both B and C.
class A:
    def show_A(self):
        print("A")

class B(A):
    def show_B(self):
        print("B")

class C:
    def show_C(self):
        print("C")

class D(B,C):
    def show_D(self):
        print("D")
# creating object:
d = D()

    # access the method
d.show_A()
d.show_B()
d.show_C()
d.show_D()
```

```python
# 2. Create Person, Employee, Student, and TeachingAssistant using hybrid inheritance.
# Real-world Hybrid Inheritance
        # What to do
# • Person is the base class.
# • Employee inherits Person.
# • Student inherits Person.
# • TeachingAssistant inherits Employee and Student.
class person:
    def __init__(self,name):
        self.name = name

class Employee(person):
    def __init__(self,name,salary):
        person.__init__(self,name)
        self.salary = salary

class Student(person):
    def __init__(self,name,marks):
        person.__init__(self,name)
        self.marks = marks

class  TeachingAssistant(Employee,Student):
    def __init__(self,name,salary,marks,department):
        Employee.__init__(self,name,salary)
        Student.__init__(self,name,marks)
        self.department = department

    def display(self):
        print("Name      :", self.name)
        print("Salary    :", self.salary)
        print("Marks     :", self.marks)
        print("Department:", self.department)

teacher = TeachingAssistant("Santosh", 50000, 90, "Data Science")

teacher.display()
```

```python
# 3. Create Vehicle → Car, Electric → ElectricCar (ElectricCar inherits from both).
        # Concept
# Hybrid Inheritance
# What to do
# • Vehicle is parent.
# • Car inherits Vehicle.
# • Electric is another parent.
# • ElectricCar inherits Car and Electric.
class Vehicle:
    def show_Vehicle(self):
        print("Vehicle is started")

class Car(Vehicle):
    def show_Car(self):
        print("Car Driving")

class Electric:
    def show_Electric(self):
        print("Electric Battery is charging")

class ElectricCar(Car,Electric):
    def show_Electric_Car(self):
        print(f"Electric car is running")
# creating object:
electriccar = ElectricCar()

    # access the method
electriccar.show_Vehicle()
electriccar.show_Car()
electriccar.show_Electric()
electriccar.show_Electric_Car()
```

```python
# 4. Demonstrate Hybrid Inheritance and print Method Resolution Order (MRO).
    # Concept
# Method Resolution Order (MRO)
# What to do
# Create
class A:
    pass

class B(A):
    pass

class C(A):
    pass

class D(B, C):
    pass

for cls in D.mro():
    print(cls.__name__)
```

```python
#6 create a banking system using hybrid inheritance 
# Account -> saving -> loan -> savingsloanAccount

class Account:
    def __init__(self,account_no):
        self.account_no = account_no

class Saving(Account):
    def __init__(self, account_no,balance):
        super().__init__(account_no)
        self.balance = balance


class Loan:
    def __init__(self, account_no, loan_amount):
        self.loan_amount = loan_amount


class SavingsLoanAccount(Saving, Loan):
    def __init__(self, account_no, balance, loan_amount):
        Saving.__init__(self, account_no, balance)
        Loan.__init__(self, account_no, loan_amount)

    def display(self):
        print("Account Number :", self.account_no)
        print("Balance        :", self.balance)
        print("Loan Amount    :", self.loan_amount)


# Create Object
obj = SavingsLoanAccount(123456789, 50000, 200000)

# Display Details
obj.display()
    
```
