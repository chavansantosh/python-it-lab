```python
# 1.Create a parent class named Person and a child class named Student.
class person:
    def __init__(self,name,age):
        self.Name = name
        self.Age = age

class Student(person):
    def __init__(self,name,age,marks):
        super().__init__(name,age)
        self.Marks = marks


    def display(self):
        print("Name :",self.Name)
        print("Age :",self.Age)
        print("Marks :",self.Marks)


# creating object
s1 = Student("santosh",24,70)

# calling method
s1.display()
```

```python
#Create a parent class named Vehicle.
    #Explanation
#The Vehicle class should contain a method:
#• start()
#Create a child class Car that:
#• Adds a brand attribute.
#• Overrides the start() method to display a car-specific starting message.
class Vehicle:
    def start(self):
        print("Vehicle is starting.")


class Car(Vehicle):
    def __init__(self, brand):
        self.brand = brand

    def start(self):
        print(self.brand, "car is starting.")


# Creating object
c1 = Car("Toyota")

# Calling method
c1.start()
```

```python
# 4. Class Name: Employee → Manager
#Create a parent class named Employee.
   #Explanation
# The Employee class should store:
#• emp_id
#• salary
#Create child class Manager.
#Add:
#• department
#Display all employee information.
class Employee:
    def __init__(self, name, emp_id, salary):
        self.Name = name
        self.Emp_Id = emp_id
        self.Salary = salary


class Manager(Employee):
    def __init__(self, name, emp_id, salary, department):
        super().__init__(name, emp_id, salary)
        self.Department = department

    def display_info(self):
        print("Name:", self.Name)
        print("Emp_Id:", self.Emp_Id)
        print("Salary:", self.Salary)
        print("Department:", self.Department)


emp1 = Manager("Santosh", 1210054, 75000, "Testing")
emp1.display_info()
```

```python
# 5. Class Name: Shape → Circle
#Create a parent class named Shape.
   #Explanation
# Create a method:
#• area()
#Create child class:
#• Circle Store:
#• radius
#Override area() to calculate the area of a circle.
class shape:
    def area(self):
        pass
class Circle(shape):
    def __init__(self,radius):
        self.radius = radius

    def area(self):
        area = 3.14 * (self.radius ** 2)
        print("Area of circle :",area)

c1 = Circle (7)
c1.area()
```

```python
# 6. Class Name: BankAccount → SavingsAccount
#Create a parent class named BankAccount.
        #Explanation
# Provide methods:
#• deposit()
#• withdraw()
#Create subclass:
#• SavingsAccount
           #Add:
#• calculate_interest()
class BankAccount:
    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print("Deposited:", amount)
        print("Balance:", self.balance)

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print("Withdrawn:", amount)
            print("Balance:", self.balance)
        else:
            print("Insufficient Balance")


class SavingsAccount(BankAccount):
    def calculate_interest(self, rate):
        interest = (self.balance * rate) / 100
        print("Interest:", interest)


# Creating object
s = SavingsAccount(10000)

s.deposit(7000)
s.withdraw(500)
s.calculate_interest(12)
```
