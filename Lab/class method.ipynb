# Basic Class & Object (1-8)

```python
#1. Create a class Student with instance attributes name and age. Create two objects and print their details.
class Student:
    def __init__(self,name,age):
        self.name = name
        self.age = age

Student1 = Student("rahit", 20)
Student2 = Student("santosh", 21)
print("Student 1")
print("Name:", Student1.name)
print("Age:", Student1.age)
print("Student 2")
print("Name:", Student2.name)
print("Age:", Student2.age)
```

```python
#2. Define class Car with attributes brand, model, and year. Create an object and display all attributes.
class Car:
    def __init__(self,brand,model,year):
        self._brand = brand
        self._model = model
        self._year = year

Car1 = Car("TATA","maruti",2023)
Car2 = Car("TATA","nano",2024)
Car3 = Car("TATA","jagwar",2026)
print("Car1")

print("Brand:",Car1._brand)
print("Model:",Car1._model)
print("Year:",Car1._year)

print("\nCar2")

print("Brand:",Car2._brand)
print("Model:",Car2._model)
print("Year:",Car2._year)


print("\nCar3")

print("Brand:",Car3._brand)
print("Model:",Car3._model)
print("Year:",Car3._year)

```

```python
#3. Create class Book with attributes title, author, price. Add a method display_info().
class Book:
    def __init__(self,title,author,price):
        self.title = title
        self.author = author
        self.price = price
    def display_info(self):
        print("Title :", self.title)
        print("Author:", self.author)
        print("Price :", self.price)
            
            
            
book = Book("Rich Dad Poor Dad", "rambabu", 99)
book.display_info()
```

```python
# 4. Create class Employee with attributes emp_id, name, salary. Create 3 objects.
class Employee:
    def __init__(self,emp_id,name,salary):
        self._emp_id = emp_id
        self._emp_name = name
        self._emp_salary = salary
        
employee1 = Employee(12345,"mahesh",20548)
employee2 = Employee(12358,"mahesh babu",20220)
        
print("employe 1")
print("emp_id",employee1._emp_id)
print("emp_name",employee1._emp_name)
print("emp_salary",employee1._emp_salary)
        
print("\nemployee 2")
print("emp_id",employee2._emp_id)
print("emp_name",employee2._emp_name)
print("emp_salary",employee2._emp_salary)
```

```python
#6. Create class Rectangle with attributes length and width. Add method to calculate area.
class Rectangle:
    def __init__(self,length,width):
        self.length = length
        self.width = width
    def area(self):
            
        return self.length * self.width

rect = Rectangle(3, 5)
            


print("Area:", rect.area())
```

```python
#8. Create class Laptop and show how to create multiple objects with different values.
class Laptop:
    def __init__(self,brand,ram,price):
        self.brand = brand
        self.ram = ram
        self.price = price

        
    def display_info(self):
        print(f"Brand :{self.brand},  Ram :{self.ram} GB,  Price $ :{self.price}")
        
laptop1 = Laptop("apple",8,125401)
laptop2 = Laptop("mango",16,4545)
laptop3 = Laptop("banana",6,55444)
laptop1.display_info()
laptop2.display_info()
laptop3.display_info()
```

```python
#9. Create class Student with class attribute school_name = "City School". Print it using class name and object.
class Student:
    def __init__(self,name,city,fee):
        self.name = name
        self.city = city
        self.fee = fee

    def display_info(self):
        print(f"Name : {self.name}, \nCity : {self.city}, \nFee $ : {self.fee}")

school = Student("City school","hyd",520414)

school.display_info()
```
