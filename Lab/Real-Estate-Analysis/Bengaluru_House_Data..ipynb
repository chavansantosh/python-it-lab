```python
# Remove duplicate properties

df['location'] = df['location'].fillna('Unknown')
df.dropna(subset=['location'], inplace=True)
df['location'].isna().sum()
```

```python
# • Standardize locations
df['location'] = df['location'].str.strip().str.title()
df['location'].unique()
```

```python
df['size'] = df['size'].fillna('Unknown')
df.dropna(subset=['size'], inplace=True)
df['size'].isna().sum()
```

```python
df['society'].isna().sum()
```

```python
df['size'] = df['size'].str.replace('Bedroom', 'BHK', regex=False)
df['size'] = df['size'].str.replace('Unknown', ' ', regex=False)
df['size'].unique()
```

```python
df = df.dropna(subset=['society'])
df['society'] = df['society'].fillna('Unknown')
df['society'].isna().sum()
```

```python
df.isnull().sum()
```

```python
df['total_sqft'].isna().sum()
```

```python
#  Extract area from 1,250 sqft

df = df.dropna(subset=['total_sqft'])
df['total_sqft'] = df['total_sqft'].fillna('Unknown')
df['total_sqft'].isna().sum()
```

```python
df['total_sqft'] = pd.to_numeric(df['total_sqft'], errors='coerce')
df['total_sqft'].dtype
```

```python
df['bath'].unique()
```

```python
# Handle missing bedrooms/bathrooms
df = df.dropna(subset=['bath'])
df['bath'] = df['bath'].fillna('Unknown')
df['bath'].isna().sum()
```

```python
df['balcony'].isnull().sum()
```

```python
df = df.dropna(subset=['balcony'])
df['balcony'] = df['balcony'].fillna('Unknown')
df['balcony'].isnull().sum()
```

```python
df['price'].isnull().sum()
```

```python
# Convert ₹85 Lakh, ₹1.2 Cr into numeric values
df['price'] = df['price'].astype(str)

df['price'] = df['price'].str.replace('₹', '', regex=False)
df['price'] = df['price'].str.replace(',', '', regex=False)
```

```python
def convert_price(x):
    if 'Lakh' in x:
        return float(x.replace('Lakh', '')) * 100000
    elif 'Cr' in x:
        return float(x.replace('Cr', '')) * 10000000
    else:
        return float(x)

df['price'] = df['price'].astype(str)
df['price'] = df['price'].apply(convert_price)
```

```python
# Detect price outliers
Q1 = df['price'].quantile(0.25)
Q3 = df['price'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

price_outliers = df[(df['price'] < lower) | (df['price'] > upper)]

price_outliers
```

```python
# Detect area outliers
Q1 = df['total_sqft'].quantile(0.25)
Q3 = df['total_sqft'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

area_outliers = df[(df['total_sqft'] < lower) | (df['total_sqft'] > upper)]

area_outliers
```
