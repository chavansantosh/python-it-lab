```python
import seaborn as sns
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("C:\\Users\\MY PC\\Downloads\\Bengaluru_House_Data (1).xls")
df.head(5)
```

```python
#  Price per sqft by location?
df
location_price = (
    df.groupby('location', as_index=False)['price_per_sqft']
      .mean()
      .sort_values('price_per_sqft', ascending=False)
      .head(10)
)

sns.barplot(
    data=location_price,
    x='price_per_sqft',
    y='location'
)
```

```python
#  Does number of bedrooms affect price?
bedroom_price = (
    df.groupby('size', as_index=False)['price']
      .mean()
      .sort_values('size')
)

sns.pointplot(
    data=bedroom_price,
    x='size',
    y='price'
)
plt.title("Number of bedrooms affect price")
```
