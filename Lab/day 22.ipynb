# Basic File Handling (1-10)

```python
# 1. Write a program to create a text file and write your name and age into it.
file = open("information.txt","w") 
file.write("name\t age\n")
file.write("santosh\t 23\n")
file.close()

file = open("information.txt","r")
print(file.read())
```

```python
#2. Read the content of a text file and print it on the screen.
with open("information.txt","w") as file:
    
    file.write("Hello,  Santosh\n")
    file.write("Welcome to  python programing")
    #file.close()
with open("information.txt","r") as file:
    content = file.read()

print(content)

print("program run successfully")
```

```python
#3. Write a program to append some text to an existing file.
with open("information.txt", "a") as file:
    file.write("\nSantosh")
    file.write("\nChavan")

with open("information.txt", "r") as file:
    content = file.read()

print(content)
```

```python
# 4. Count the number of lines in a text file.
with open("information.txt", "r") as file:
    lines = file.readlines()

print("number of lines:",len(lines))
```

```python
# 5. Count the number of words in a text file.

with open("information.txt", "w") as file:
    file.write("Hello Santosh\n")
    file.write("welcome to python programming")
    
with open("information.txt", "r") as file:
    
    content = file.read()
    words = content.split()
    
print("number of words:",len(words))
```

```python
#6. Count the number of characters in a text file.
with open("information.txt", "r") as file:
    content = file.read()
    
print("number of character:",len(content))
```

```python
#7. Read a file line by line using a loop.
with open("information.txt", "r") as file:
    for line in file:
        print(line,end=" ")
```

```python
#8. Write a program to copy content from one file to another file.
with open("information.txt", "r") as file:
    content = file.read()
    
with open("santo.txt","w") as file:
    file.write(content)

print("copied")

with open("santo.txt","r") as file:
    
    print(file.read())
```

# Intermediate File Handling (11-20)

```python
# 11. Take student name and marks from user and store them in a file.
with open("inform.txt","w") as file:
    file.write(input("Enter your name:"))
    file.write(input("Enter your marks:"))

with open("inform.txt","r") as file:
    print(file.read())
```

```python
#12. Read a file and print only the lines that contain the word “Python”.
with open("inform.txt","w") as file:
    file.write("Hi This is\n")
    file.write("python file\n")
    file.write("\nprograming language")
    
with open("inform.txt","r") as file:
    for line in file:

        if "python" in line.lower():
            print(line,end=" ")
```

```python
#13. Write a program to reverse the content of a file and save it in another file.
with open("inform.txt","w") as file:
    file.write("hi this is\n")
    file.write("python language\n")
    file.write("programing\n")
with open("inform.txt","r") as file:
    content = file.read()
    reverse_content = content[::-1]

with open("inform.txt","w") as file:
    
    file.write(reverse_content)

   # for line in file:
      #  if i in content:
          #  print(i.reverse())
with open("inform.txt","r") as file:
    print(file.read())
```

```python
#14. Count how many times a particular word appears in a file.
with open("inform.txt","r") as file:
    content = file.read().lower()
    count = content.count("python")
    print(count)
```

```python
#15. Read a file and display the longest word in it.
with open("inform.txt","w") as file:
    file.write("hi this is\n")
    file.write("python language\n")
    file.write("programing\n")
    
with open("inform.txt","r") as file:
    
    content = file.read()
    words = content.split()

longest_word = max(words,key=len)

print(longest_word)
```
