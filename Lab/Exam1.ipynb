```python
""" 1. Custom Exceptions
Create the following custom exceptions:
• InsufficientBalanceError → Raised when withdrawal amount is greater than balance
• InvalidAmountError → Raised when amount is zero or negative
• AccountNotFoundError → Raised when account number does not exist
• DailyLimitExceededError → Raised when daily withdrawal limit is crossed
"""
class InsufficientBalanceError(Exception):
    pass


class InvalidAmountError(Exception):
    pass


class AccountNotFoundError(Exception):
    pass


class DailyLimitExceededError(Exception):
    pass


def withdraw(balance, amount, account_exists, daily_limit):

    if amount > balance:
        raise InsufficientBalanceError("Insufficient balance")

    if amount <= 0:
        raise InvalidAmountError("Amount should be more than zero")

    if not account_exists:
        raise AccountNotFoundError("Account not found")

    if amount > daily_limit:
        raise DailyLimitExceededError("Your daily limit exceeded")

    balance -= amount
    return balance


balance = 25000
amount = 4000
account_exists = True
daily_limit = 24000

try:
    new_balance = withdraw(balance, amount, account_exists, daily_limit)
    print("Withdrawal Successful")
    print("Remaining Balance:", new_balance)

except InsufficientBalanceError as e:
    print(e)

except AccountNotFoundError as e:
    print(e)

except InvalidAmountError as e:
    print(e)

except DailyLimitExceededError as e:
    print(e)
```

```python
"""2. Abstract Class
Create an abstract class Account with:
• Abstract methods: deposit(amount), withdraw(amount), get_balance()
• Concrete method: show_account_info()
"""
from abc import ABC, abstractmethod


class Account(ABC):

    @abstractmethod
    def deposit(self, amount):
        pass

    @abstractmethod
    def withdraw(self, amount):
        pass

    @abstractmethod
    def get_balance(self):
        pass

    def show_account_info(self):
        print("This is a bank account")


class SavingsAccount(Account):

    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
        else:
            print("Insufficient balance")

    def get_balance(self):
        return self.balance


class CurrentAccount(Account):

    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
        else:
            print("Insufficient balance")

    def get_balance(self):
        return self.balance


savingaccount = SavingsAccount(5000)
currentaccount = CurrentAccount(70000)

savingaccount.deposit(2000)
savingaccount.withdraw(1000)

print("Savings Account Balance:", savingaccount.get_balance())
savingaccount.show_account_info()

currentaccount.deposit(5000)
currentaccount.withdraw(2000)

print("Current Account Balance:", currentaccount.get_balance())
currentaccount.show_account_info()
```

```python
""" 3. Concrete Classes
A. SavingsAccount (inherits from Account)
• Private attributes: __account_no, __name, __balance, __daily_withdrawn
• Daily withdrawal limit = ₹25,000
• Minimum balance = ₹1,000
• Interest rate = 4%
B. CurrentAccount (inherits from Account)
• Private attributes: __account_no, __name, __balance
• Overdraft limit = ₹10,000
• No interest
"""
from abc import ABC, abstractmethod


class Account(ABC):

    @abstractmethod
    def deposit(self, amount):
        pass

    @abstractmethod
    def withdraw(self, amount):
        pass

    @abstractmethod
    def get_balance(self):
        pass

    def show_account_info(self):
        print("Account Information")


class SavingsAccount(Account):

    DAILY_WITHDRAWAL_LIMIT = 25000
    MINIMUM_BALANCE = 1000
    INTEREST_RATE = 0.04

    def __init__(self, account_no, name, balance):
        self.__account_no = account_no
        self.__name = name
        self.__balance = balance
        self.__daily_withdrawn = 0

    def deposit(self, amount):
        if amount <= 0:
            print("Invalid deposit amount")
        else:
            self.__balance += amount
            print("Deposit successful")

    def withdraw(self, amount):
        if amount <= 0:
            print("Invalid withdrawal amount")

        elif self.__daily_withdrawn + amount > self.DAILY_WITHDRAWAL_LIMIT:
            print("Daily withdrawal limit exceeded")

        elif self.__balance - amount < self.MINIMUM_BALANCE:
            print("Minimum balance of ₹1000 must be maintained")

        else:
            self.__balance -= amount
            self.__daily_withdrawn += amount
            print("Withdrawal successful")

    def get_balance(self):
        return self.__balance

    def calculate_interest(self):
        return self.__balance * self.INTEREST_RATE

    def show_account_info(self):
        print("Account Number:", self.__account_no)
        print("Name:", self.__name)
        print("Balance:", self.__balance)
        print("Daily Withdrawn:", self.__daily_withdrawn)
        print("Interest Rate:", self.INTEREST_RATE * 100, "%")


class CurrentAccount(Account):

    OVERDRAFT_LIMIT = 10000

    def __init__(self, account_no, name, balance):
        self.__account_no = account_no
        self.__name = name
        self.__balance = balance

    def deposit(self, amount):
        if amount <= 0:
            print("Invalid deposit amount")
        else:
            self.__balance += amount
            print("Deposit successful")

    def withdraw(self, amount):
        if amount <= 0:
            print("Invalid withdrawal amount")

        elif amount > self.__balance + self.OVERDRAFT_LIMIT:
            print("Overdraft limit exceeded")

        else:
            self.__balance -= amount
            print("Withdrawal successful")

    def get_balance(self):
        return self.__balance

    def show_account_info(self):
        print("Account Number:", self.__account_no)
        print("Name:", self.__name)
        print("Balance:", self.__balance)
        print("Overdraft Limit:", self.OVERDRAFT_LIMIT)
        print("Interest: No interest")


# Creating objects

saving = SavingsAccount("SA101", "Rahul", 50000)

current = CurrentAccount("CA101", "Santosh", 70000)


# Savings Account

saving.show_account_info()

saving.deposit(5000)
saving.withdraw(10000)

print("Savings Balance:", saving.get_balance())
print("Interest:", saving.calculate_interest())


print("------------------------")


# Current Account

current.show_account_info()

current.deposit(5000)
current.withdraw(75000)

print("Current Balance:", current.get_balance())
```
