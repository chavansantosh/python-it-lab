# Positional Arguments & Keyword Arguments

```python
# 13. Write a function full_name(first, last) and call it using positional arguments.
def full_name(first, last):
    return f"first name  {first} and last {last}"
```

```python
full_name("santosh","chavan")
```

```python
#14. Call the same function using keyword arguments.
def full_name(first,last):
    return f"name {first}  {last}"

print(full_name(first="santosh", last="chavan"))
print(full_name(first="chavan", last="santosh"))
```

```python
#15. Write a function details(name, age, city) and demonstrate both positional and keyword calls.
def details(name, age, city):
    return f" my name is {name} and age {age},i live in city {city}"

print(details(name="santosh",age=25,city="hyd"))
print(details("santosh",25,"hyd"))
```

```python
#16. Create a function calculate(a, b, operation) and show how keyword args improve readability.
def calculate(a, b, operation):
    if operation == "add":
        return a + b
    elif operation == "substract":
        return a - b
    elif operation == "multiply":
        return a * b
    elif operation == "divide":
        return a / b

print(calculate(a=4, b=7, operation="add"))
```
