```python
# 11.Rename the following columns: CustomerId → Customer_ID, CreditScore → Credit_Score, EstimatedSalary → Salary, Exited → Churn
df.rename(columns = {"CustomerId":"CustomerID","CreditScore":"Credit_Score","EstimatedSalary":"Salary","Exited":"churn"},inplace = True)
df
```

```python
# 12.Add the prefix Bank_to every column name.
df=df.add_prefix("Bank_")
df
```

```python
# 13.Convert the Age column into float datatype.
df['Age']=df['Bank_Bank_Age'].astype(float)
df["Age"]
#df.dtypes
```

```python
# 14.Convert the Exited column into string datatype.
df['Bank_Bank_churn']=df['Bank_Bank_churn'].astype(str)
df["Bank_Bank_churn"]
```

```python
# 15.Suppose the Balance column has missing values. Replace them using the mean of the Balance column.
df['Bank_Bank_Balance'] = df['Bank_Bank_Balance'].fillna(df['Bank_Bank_Balance']).mean()
print(df['Bank_Bank_Balance'].head())
```

```python
# 16.Suppose the Age column has missing values.
# Replace them using the median age.
df["Bank_Bank_Age"] = df["Bank_Bank_Age"].fillna(df["Bank_Bank_Age"].median())
df["Bank_Bank_Age"]
```
