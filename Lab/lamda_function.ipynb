# Easy Level (1–10)

```python
# 1.Write a lambda function to return the square of a number. 
num = lambda x: x*x
num(5)
```

```python
# 2.Write a lambda function to return the cube of a number.
cube = lambda x: pow(x,3)
cube(2)
```

```python
#3.Write a lambda function to add two numbers.
add = lambda x,y: x+y
add(7,5)
```

```python
# 4.Write a lambda function to find the maximum of two numbers.
maximum = lambda x,y : x if x > y else y
print(maximum(7,5))
print(maximum(46,68))
```

```python
# 5.Write a lambda function that returns "Even" if a number is even; otherwise "Odd".
even_odd = lambda x,y: "even" if x % 2 == 0 else "odd"
print(even_odd(8,8))
```

```python
# 6.Use map() with lambda to square every element of a list.
number = [1,2,3,4]
num =list(map(lambda x: x*x,number))
num
```

```python
# 7.Use filter() with lambda to keep only even numbers.
number = [1,2,3,4,5,6,7,]
result = list(filter(lambda x: x % 2 == 0, number ))
result
```

```python
# 8.Use filter() with lambda to keep numbers greater than 50.
numbers = [20,52,87,63]
result = list(filter(lambda x: x > 50 ,numbers))
result
```

```python
# 9.Sort a list of numbers in ascending order using lambda.
numbers = [50, 20, 40, 10, 30]

numbers.sort(key=lambda x: x)

print(numbers)
```

```python
#10.Sort a list of numbers in descending order using lambda.
numbers = [10, 20, 4, 1, 30]

numbers.sort(key=lambda x: -x)

print(numbers)
```

# Medium Level (11–20)

```python
# 11.Use reduce() with lambda to find the sum of all numbers.[10,20,30,40]
from functools import reduce
numbers = [10,20,30,40]
result = reduce(lambda x,y: x +y, numbers)
print(numbers)
```

```python
# 12.Use reduce() with lambda to find the product of all numbers. [2,3,4,5]
from functools import reduce
numbers = [2,3,4,5]
result = reduce(lambda x,y: x*y,numbers)
print(result)
```

```python
# 13.Sort the following list according to age.
students = [("Ram",22),("Amit",19),("Neha",25),("Raj",20)]
sorted_students =sorted(students,key=lambda students: students[1])
print(sorted_students)
```

```python
# 14.Sort the following dictionary by values.
marks = {"A":80,"B":95,"C":70,"D":88}
sorted_marks =dict(sorted(marks.items(),key=lambda items: items[1]))
print(sorted_marks)
```

```python
# 15.Find the longest string using lambda.
fruits =["apple","banana","kiwi","watermelon"]
largests = max(fruits,key=lambda x: x)
print(largests)
```

```python
# 16.Use lambda with max() to find the student having the highest marks.
students = [("Ram",85),("Amit",91),("Neha",78)]
highest_marks = max(students,key=lambda x: x[1])
print(highest_marks)                    
```

```python
#17.Use lambda with min() to find the employee with the lowest salary.
employees = [("A",35000),("B",42000),("C",28000)]
lowest_salary= min(employees,key=lambda x:x[1])
print(lowest_salary)
```
