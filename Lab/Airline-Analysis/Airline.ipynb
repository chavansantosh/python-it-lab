```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
df =pd.read_csv("C:\\Users\\MY PC\\Downloads\\Airline_info.xls")
df
```

```python
# Which airline has the highest average price?

avg_price = df.groupby('Airline')['Price'].mean().sort_values(ascending=False)
```

```python
avg_price.plot(kind='bar', figsize=(10,5))
plt.xlabel('Airline')
plt.ylabel('Average Price')
plt.title('Average Price by Airline')
plt.xticks(rotation=45)
plt.show()
```

```python
# Non-stop vs connecting flight prices?
df.groupby('Total_Stops')['Price'].mean()
```

```python
df.groupby('Total_Stops')['Price'].mean().plot(kind='bar')

plt.xlabel('Stops')
plt.ylabel('Non-stop')
plt.title('Non-stop vs Connecting Flight Prices')
plt.show()
```

```python
df.groupby('Destination')['Price'].mean()
```

```python
df.groupby('Destination')['Price'].mean().plot(kind='bar')

plt.xlabel('Stops')
plt.ylabel('expensive Price')
plt.title('Destination vs expensive Price')
plt.show()
```

```python
#  Does duration affect price?

df.groupby('Duration')['Price'].mean().plot(kind='line')

plt.xlabel('Duration (minutes)')
plt.ylabel('affect Price')
plt.title('Duration vs affect Price')
plt.show()
```

```python
#  Which departure times have higher prices?
df.groupby('Dep_Time')['Price'].mean().plot(kind='line')

plt.xlabel('Dep_Time')
plt.ylabel('higher Price')
plt.title('Dep_Time vs higher Price')
plt.show()
```
