# Seaborn catplot() —  Questions

```python
import seaborn as sns
import numpy as np
import pandas as pd
df = pd
import matplotlib.pyplot as plt
titanic = sns.load_dataset("titanic")
titanic
```

## 1. Survival by Gender
## Use catplot() to show the number of passengers who survived and did not survive for each
## gender.
## • X-axis → sex
## • Hue → survived
## • Chart → count

```python
sns.catplot(data = titanic,
            x = "sex",
           # hue = "survived",
            kind = "count"
           )
```

## 7. Fare by Passenger Class
## Use catplot() to compare fare across the three passenger classes.
## Add:
## • Appropriate color
## • Title
## • X/Y labels

```python
sns.catplot(data =titanic,
            x = 'pclass',
            y = 'fare',
            palette = 'dark',
            kind = 'bar'
    
)
plt.title('Age Distribution by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('fare')
plt.show()
```

## 14. Class, Gender and Survival
## Create a categorical visualization where:
## • X → class
## • Y → survival-related measure
## • Hue → sex
## • Columns → survived
## Use col= to create separate panels.

```python
sns.catplot(
    data=titanic,
    x='class',
    hue='sex',
    col='survived',
    kind='count'
)

plt.show()
```

## 15. Titanic Multi-Panel Analysis
## Create a catplot() with:
## • X → class
## • Y → fare
## • Hue → sex
## • Columns → survived

```python
sns.catplot(
    data=titanic,
    x='class',
    y='fare',
    hue='sex',
    col='survived',
    kind='bar',
    height=5,
    aspect=1.2,
    palette='Set2'
)
plt.title("Titanic Multi-Panel Analysis",y=1.05)
plt.xlabel("class")
plt.ylabel("fare")
plt.show()
```
