# Easy Level (1-7)

```python
#1.Take a string from user and replace all spaces with underscore _ using replace() method.
name =str(input("please enter a string:"))
print(name.replace(" ","--"))
```

```python
# 2.Take any string and remove leading and trailing spaces using strip().
s ="   python   "
print(s.strip())
```

```python
# 3.Take a sentence and check whether it starts with "Hello" using startswith().
name = "Hello im babu"

print(name.startswith("Hello"))
```

```python
# 4.Take a string and check if it ends with ".com" using endswith().
S = "paritosh@gmail.com"
print(S.endswith(".com"))
```

```python
# 5.Take a comma-separated string like "apple,banana,mango" and split it using split().
S = "apple,banana,mango"

print(S.split(","))
```

```python
# 6.Create a list of 5 fruits and join them using space " " with join().
fruits = ["mango","banana","orange","guava","grapees"]
print(" ".join(fruits))
```

```python
# 7.Create a list: [" Python ", " is ", " awesome "]. Strip each word and join them.
list1 =["python","is","awesome"]
A =list1[0].strip()
B =list1[1].strip()
C =list1[2].strip()
D = "  ".join([A,B,C])
print(D)
```
