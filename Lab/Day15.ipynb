# Basic Multiple Inheritance (1-10)

```python
# 1. Create two parent classes Father and Mother. Create a child class Child that inherits from both.
# Parent class Father
class Father:
    def __init__(self):
        self._father_name = "ABc"
    
    def father_info(self):
        return f"father's name is {self._father_name}"

# Parent class Mother
class Mother:
    def __init__(self):
        self._mother_name = "KCPD"
    
    def mother_info(self):
        return f"mother's name is {self._mother_name}"

# Child class
class Child(Father, Mother):
    def __init__(self):
        
        # Call constructors of both parents
        Father.__init__(self)
        Mother.__init__(self)
    
    def child_info(self):
        return "child inherited properties of father and mother"
#calling function

child = Child()
print(child.father_info()) 
print(child.mother_info())  
print(child.child_info())   
```

```python
# 2. Define class A with method show() and class B with method display(). Create class C that inherits from both and calls both methods.
class A:
    
    def show(self):
        return "A"

class B:
    def display(self):
        return "B"

class C(A,B):

    def method(self):
        print(super().show())
        print(super().display())

c = C()
c.method()
```

```python
# 3. Create classes Teacher and Author. Create class Person that inherits from both and has attributes from both.
class Teacher:
    def __init__(self):
        self._teacher_name = "paritosh"

    def teacher_info(self):
        return f"Teacher name is: {self._teacher_name}"

class Author:
    def __init__(self):
        self._author_name = "santosh"

    def author_info(self):
        return f"Author name is : {self._author_name}"

# child class
class person(Teacher,Author):
    def __init__(self):


    # calling constructor from parents
         Teacher.__init__(self)
         Author.__init__(self)

    def person_info(self):
        return f"Name of Teacher {self._teacher_name} and Author {self._author_name}"

    
p1 = person()
print(p1.teacher_info())
print(p1.author_info())
print(p1.person_info())
```

```python
# 4. Define class Sports and class Academic. Create class Student that inherits from both.
class Sport:
    def __init__(self):
        self._sport_name = "criket"

    def sport_info(self):
        return f"World famous sport is : {self._sport_name}"

class Acedemic:
    def __init__(self):
        self._acedemic = "good"

    def acedemic_info(self):
        return f"Acedenically very :{self._acedemic}"

class Student(Sport,Acedemic):
    def __init__(self):

        #calling constructor from parents
        Sport.__init__(self)
        Acedemic.__init__(self)

    def Student_info(self):
        return f"Name of Sport {self._sport_name} and Acedemic {self._acedemic}"


        
s1 = Student()
print(s1.sport_info())
print(s1.acedemic_info())
print(s1.Student_info())
```
